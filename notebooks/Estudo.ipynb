```python
# Exercício 1

print("Seja bem-vinda à Empresa X da Beatriz Toledo!")
valor_unitario = float(input("Digite o valor unitário do produto: R$ "))
quantidade = int(input("Digite a quantidade do produto: "))

# Cálculo do valor total sem desconto
valor_total = valor_unitario * quantidade

# Verificação de desconto
if valor_total < 2500:
    desconto = 0
elif valor_total >= 2500 and valor_total < 6000:
    desconto = 0.04
elif valor_total >= 6000 and valor_total < 10000:
    desconto = 0.07
else:
    desconto = 0.11

# Cálculo do valor do desconto
valor_desconto = valor_total * desconto

# Cálculo do valor total com desconto
valor_com_desconto = valor_total - valor_desconto

# Exibição dos resultados
print("\n--- RESUMO DA COMPRA ---")
print(f"Valor total sem desconto: R$ {valor_total:.2f}")
print(f"Desconto aplicado: {desconto * 100:.0f}%")
print(f"Valor do desconto: R$ {valor_desconto:.2f}")
print(f"Valor total com desconto: R$ {valor_com_desconto:.2f}")
```

```python
# Exercício 2

print("Seja bem-vinda à loja de Açaí e Cupuaçu da Beatriz Toledo!")
print("Tamanho P: CP (R$ 9), AC (R$11)"
      "Tamanho M: CP (R$ 14), AC (R$16)"
      "Tamanho G: CP (R$ 18), AC (R$20)")

# Acumulador para armazenar o valor total dos pedidos
total = 0
valor = 0

while True:

# Solicita o sabor do produto
  sabor = (input("Digite o sabor do produto (AC para Açaí/ CP para Cupuaçu): "))

# Verifica se o sabor é válido
  if sabor != "AC" and sabor != "CP":
    print("Sabor inválido. Tente novamente.")
    continue

# Solicita o tamanho do produto
  tamanho =(input("Digite o tamanho do produto (P/M/G): "))

# Verifica se o tamanho é válido
  if tamanho != "P" and tamanho != "M" and tamanho != "G":
    print("Tamanho inválido. Tente novamente.")
    continue

# Define preços
  if sabor == "CP":
        if tamanho == "P":
            valor = 9
        elif tamanho == "M":
            valor = 14
        else:
            valor = 18

  elif sabor == "AC":
        if tamanho == "P":
            valor = 11
        elif tamanho == "M":
            valor = 16
        else:
            valor = 20

# Adiciona o valor do pedido ao acumulador
  total += valor

# Mostra o valor do pedido atual
  print(f"Valor do pedido: R$ {valor:.2f}")

# Pergunta se o cliente deseja realizar outro pedido
  continuar = input("Deseja pedir mais alguma coisa? (S/N): ")

# Se não quiser continuar, encerra o while
  if continuar == "N":
        break

# Exibe o valor total dos pedidos
print(f"\nValor total dos pedidos: R$ {total:.2f}")
print("Obrigado pela preferência!")
```

```python
# Exercício 3

print("Bem-vinda à Copiadora - Beatriz Toledo!")

# Funções
def escolha_servico():
    while True:
        servico = input("Digite o serviço desejado (DIG/ICO/IPB/FOT): ")

        if servico == "DIG":
            return 1.10
        elif servico == "ICO":
            return 1.00
        elif servico == "IPB":
            return 0.40
        elif servico == "FOT":
            return 0.20
        else:
            print("Opção de serviço inválida. Tente novamente.")

def num_pagina():
    while True:
        try:
            paginas = int(input("Digite o número de páginas: "))

            if paginas >= 20000:
                print("Não aceitamos pedidos com 20000 páginas ou mais.")
                continue

            elif paginas >= 2000:
                return paginas * 0.75

            elif paginas >= 200:
                return paginas * 0.80

            elif paginas >= 20:
                return paginas * 0.85

            else:
                return paginas

        except ValueError:
            print("Valor inválido. Digite um número inteiro.")

def servico_extra():
    while True:
        extra = input(
            "Digite o serviço extra "
            "(1 - Encadernação simples / "
            "2 - Capa dura / "
            "0 - Nenhum): ")

        if extra == "1":
            return 15
        elif extra == "2":
            return 40
        elif extra == "0":
            return 0
        else:
            print("Opção de serviço extra inválida. Tente novamente.")


# Programa principal
servico = escolha_servico()
paginas = num_pagina()
extra = servico_extra()

# Calcula o valor total
total = (servico * paginas) + extra

# Exibe o resultado
print(f"\nValor total a pagar: R$ {total:.2f}")
```

```python
# Exercício 4

print("Bem-vinda ao sistema de gerenciamento de livros - Beatriz Toledo!")

lista_livros = []
id_global = 1

# Funções

def cadastrar_livro(id):
    print("\n--- CADASTRO DE LIVRO ---")

    nome = input("Digite o nome do livro: ")
    autor = input("Digite o autor do livro: ")
    editora = input("Digite a editora do livro: ")

    # Dicionário com os dados do livro
    livro = {
        "id": id,
        "nome": nome,
        "autor": autor,
        "editora": editora
    }

    # Copia o dicionário para dentro da lista
    lista_livros.append(livro.copy())

    print("Livro cadastrado com sucesso!")

def consultar_livro():

    while True:
        print("\n--- CONSULTAR LIVRO ---")
        print("1 - Consultar Todos")
        print("2 - Consultar por Id")
        print("3 - Consultar por Autor")
        print("4 - Retornar ao menu")

        opcao = input("Digite a opção desejada: ")

        # Consultar todos os livros
        if opcao == "1":
            if len(lista_livros) == 0:
                print("Não há livros cadastrados.")
            else:
                for livro in lista_livros:
                    print("\nLivro:")
                    print(f"ID: {livro['id']}")
                    print(f"Nome: {livro['nome']}")
                    print(f"Autor: {livro['autor']}")
                    print(f"Editora: {livro['editora']}")

        # Consultar livro pelo ID
        elif opcao == "2":
            id_consulta = input("Digite o ID do livro: ")

            encontrado = False

            for livro in lista_livros:
                if str(livro["id"]) == id_consulta:
                    print("\nLivro encontrado:")
                    print(f"ID: {livro['id']}")
                    print(f"Nome: {livro['nome']}")
                    print(f"Autor: {livro['autor']}")
                    print(f"Editora: {livro['editora']}")
                    encontrado = True
                    break

            if encontrado == False:
                print("Id inválido")

        # Consultar livros pelo autor
        elif opcao == "3":
            autor_consulta = input("Digite o nome do autor: ")

            encontrado = False

            for livro in lista_livros:
                if livro["autor"].lower() == autor_consulta.lower():
                    print("\nLivro encontrado:")
                    print(f"ID: {livro['id']}")
                    print(f"Nome: {livro['nome']}")
                    print(f"Autor: {livro['autor']}")
                    print(f"Editora: {livro['editora']}")
                    encontrado = True

            if encontrado == False:
                print("Autor não encontrado.")

        # Retorna para o menu principal
        elif opcao == "4":
            return

        # Opção inválida
        else:
            print("Opção inválida")


def remover_livro():

    while True:
        id_remover = input("Digite o ID do livro que deseja remover: ")

        for livro in lista_livros:
            if str(livro["id"]) == id_remover:
                lista_livros.remove(livro)
                print("Livro removido com sucesso!")
                return

        print("Id inválido")


# MENU PRINCIPAL
while True:

    print("\n--- MENU PRINCIPAL ---")
    print("1 - Cadastrar Livro")
    print("2 - Consultar Livro")
    print("3 - Remover Livro")
    print("4 - Encerrar Programa")

    opcao = input("Digite a opção desejada: ")

    # Cadastrar livro
    if opcao == "1":
        cadastrar_livro(id_global)

        id_global += 1

    # Consultar livros
    elif opcao == "2":
        consultar_livro()

    # Remover livro
    elif opcao == "3":
        remover_livro()

    # Encerrar programa
    elif opcao == "4":
        print("Programa encerrado.")
        break

    # Opção inválida
    else:
        print("Opção inválida")
```

```python
# Exercício 5

from random import randint

print("Bem-vindo ao sistema de BINGO!")

cartelas = []
sorteados = []
regra = 1 # Regra padrão: Linha, Coluna ou Diagonal

# --------------------------------------------------
# Função 1 - Gerar cartelas
# --------------------------------------------------

def gerar_cartela(sigla):
    cartela = []

    for i in range(5): # Percorre as cinco linhas da cartela
        linha = []

        for j in range(5): # Percorre as cinco colunas da linha atual

            if i == 2 and j == 2:
                linha.append(sigla)
                continue

            if j == 0:
                minimo = 1
                maximo = 15
            elif j == 1:
                minimo = 16
                maximo = 30
            elif j == 2:
                minimo = 31
                maximo = 45
            elif j == 3:
                minimo = 46
                maximo = 60
            else:
                minimo = 61
                maximo = 75

            while True:
                numero = randint(minimo, maximo) # Gera um número inteiro dentro do intervalo da coluna atual
                repetido = False

                # CORREÇÃO: Confere apenas as linhas que já foram adicionadas à cartela (de 0 até i-1)
                for k in range(i):
                    if cartela[k][j] == numero:
                        repetido = True

                if repetido == False:
                    linha.append(numero)
                    break

        cartela.append(linha)

    return cartela

# --------------------------------------------------
# Função 2 - Imprimir tabela
# --------------------------------------------------

def imprimir_cartela(cartela):

    for linha in cartela: # Percorre cada linha da cartela

        for valor in linha:  # Percorre cada valor da linha atual

            if isinstance(valor, int) and valor < 10:
                print(f"[0{valor}]", end=" ") # Coloca zero na frente dos números menores que 10

            else:
                print(f"[{valor}]", end=" ")

        # Pula para a próxima linha
        print()

# --------------------------------------------------
# Função 3 - Sortear valor
# --------------------------------------------------
def sorteia_valor(sorteados):

    if len(sorteados) >= 75:
        return -1 # Se todos os 75 números já foram sorteados

    while True: # Continua sorteando até encontrar um número novo

        numero = randint(1, 75)

        if numero not in sorteados:
            return numero

# --------------------------------------------------
# Função auxiliar
# --------------------------------------------------

def verifica_sequencia(sequencia, sorteados):

    ganhou = True

    for valor in sequencia:

        if isinstance(valor, int) and valor not in sorteados:
            ganhou = False

    return ganhou

# --------------------------------------------------
# Função 4 - Verificar ganhador
# --------------------------------------------------
def verifica_ganhador_cheia(cartelas, sorteados):

    for cartela in cartelas: # Verifica cada cartela

        completou_todas = True

        for linha in cartela: # Percorre cada linha

            if not verifica_sequencia(linha, sorteados):
                completou_todas = False
                break

        if completou_todas:
            return cartela # Se todas as linhas foram preenchidas

    return None # Se não encontrou nenhuma cartela vencedora

# --------------------------------------------------
# Função 5 - Verificar linha, coluna e diagonal
# --------------------------------------------------
def verifica_ganhador_LCD(cartelas, sorteados):

    for cartela in cartelas:

        # VERIFICANDO AS LINHAS
        for linha in range(5):

            if verifica_sequencia(cartela[linha], sorteados):
                return cartela

        # VERIFICANDO AS COLUNAS
        for coluna in range(5):

            sequencia = []

            for linha in range(5):
                sequencia.append(cartela[linha][coluna])

            if verifica_sequencia(sequencia, sorteados):
                return cartela

        # DIAGONAL PRINCIPAL
        sequencia = []

        for i in range(5):
            sequencia.append(cartela[i][i])

        if verifica_sequencia(sequencia, sorteados):
            return cartela

        # DIAGONAL SECUNDÁRIA
        sequencia = []

        for i in range(5):
            sequencia.append(cartela[i][4 - i])

        if verifica_sequencia(sequencia, sorteados):
            return cartela

    return None

# ==================================================
# MENU
# ==================================================

while True:

    print("\n===== MENU BINGO =====")
    print("1 - Gerar Cartelas")
    print("2 - Definir Regras")
    print("3 - Começar Bingo!")
    print("4 - Encerrar Programa")

    opcao = input("Digite a opção desejada: ")

    # OPÇÃO 1 - GERAR CARTELAS

    if opcao == "1":

        cartelas = []  # Apaga as cartelas anteriores

        quantidade = int(
            input("Quantas cartelas deseja gerar? ")
        )

        sigla = input("Digite suas iniciais: ")

        for i in range(quantidade):

            cartela = gerar_cartela(sigla)

            cartelas.append(cartela)

            print(f"\nCartela {i + 1}:")
            imprimir_cartela(cartela)

    # OPÇÃO 2 - DEFINIR REGRAS

    elif opcao == "2":

        print("\nEscolha a regra do bingo:")
        print("1 - Linha, Coluna ou Diagonal (LCD)")
        print("2 - Cartela cheia")

        regra = int(input("Digite a regra desejada: "))

        if regra != 1 and regra != 2:
            regra = 1
            print("Opção inválida. A regra LCD será utilizada.")

        else:
            print("Regra definida com sucesso!")

    # OPÇÃO 3 - COMEÇAR BINGO

    elif opcao == "3":

        if len(cartelas) == 0:  # Verifica se existem cartelas

            print("Nenhuma cartela foi gerada.")

            continue

        sorteados = [] # Começa uma nova lista de números sorteados

        print("\n===== BINGO COMEÇOU! =====")

        while True:

            numero = sorteia_valor(sorteados) # Sorteia um número

            if numero == -1:

                print("Todos os números foram sorteados.")

                break

            sorteados.append(numero) # Adiciona o número na lista

            print(f"Número sorteado: {numero}")


            # REGRA CARTELA CHEIA

            if regra == 2:

                ganhador = verifica_ganhador_cheia(
                    cartelas,
                    sorteados
                )

                if ganhador is not None: # Se encontrou um ganhador

                    print("\nCARTELA VENCEDORA!")

                    imprimir_cartela(ganhador)

                    print(
                        f"\nQuantidade de números sorteados: "
                        f"{len(sorteados)}"
                    )

                    break

            # REGRA LINHA, COLUNA OU DIAGONAL

            else:

                ganhador = verifica_ganhador_LCD(
                    cartelas,
                    sorteados
                )

                if ganhador is not None:

                    print("\nCARTELA VENCEDORA!")

                    imprimir_cartela(ganhador)

                    print(
                        f"\nQuantidade de números sorteados: "
                        f"{len(sorteados)}"
                    )

                    break

    # OPÇÃO 4 - ENCERRAR

    elif opcao == "4":

        print("Programa encerrado.")

        break

    # OPÇÃO INVÁLIDA

    else:

        print("Opção inválida!")
```
